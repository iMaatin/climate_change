# grouper v2

Reads CSVs saved by `grouper` and recodes World Bank income groups so that
**Lower middle income** and **Upper middle income** become **Middle income**.

```python
import os
import pandas as pd
import glob
```

```python

def map_income_groups(input_folder, output_folder):
    dict2 = {
        'Lower middle income': 'Middle income',
        'Upper middle income': 'Middle income',
        'High income': 'High income',
        'Low income': 'Low income'
    }

    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".csv"):
            input_path = os.path.join(input_folder, filename)
            output_path = os.path.join(output_folder, filename)

            df = pd.read_csv(input_path)

            if "income_group" not in df.columns:
                print(f"Skipping {filename}: 'income_group' column not found.")
                continue

            df["income_group"] = df["income_group"].map(dict2)

            df.to_csv(output_path, index=False)

            print(f"Saved: {output_path}")
```

```python
emissions_pre = r'D:\Prog\ICRI\climate_change\co2_calc\emissions_v4'
emissions_post = r'D:\Prog\ICRI\climate_change\co2_calc\emissions_v4.1'

pre = r'D:\Prog\ICRI\climate_change\paper_2\processed'
post = r'D:\Prog\ICRI\climate_change\paper_2\processed_v1.1'

map_income_groups(emissions_pre, emissions_post)
map_income_groups(pre, post)
```
